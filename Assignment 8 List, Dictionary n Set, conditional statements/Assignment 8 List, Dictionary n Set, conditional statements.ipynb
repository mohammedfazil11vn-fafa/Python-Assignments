# Sets (Operations):


a. Create a set called my_set with following values:
['a','e','i','o','u','a','a','i']

Analyse the output and provide explanation for the same.




```python
my_set = set(['a','e','i','o','u','a','a','i'])
print(my_set)
```

set is a collection of unique and unordered items. Sets do not allow duplicate values. Here in the given input a appears three times and i appears two times. The original input contains 8 elements but the resulting set contains only 5 unique elements. Sets are unordered ,so the output may appear in a different order each time the code runs.

b. Attempt to change the value of my_set[4] = 's'. If code throws an error, provide
an explanation.

```python
my_set[4] ="s"
```

The error occurs because the sets are unordered collections and do not support indexing. The elements of a set do not have fixed positions. The output may appear in a different order each time the code runs. Therefore, we cannot access or modify elements using the indexing method.

c. Create two sets:
    
set1 with values: {1, 3, 5, 7, 9}
    
set2 with values: {2, 3, 5, 8, 10}
    
d. Compute and print the union and intersection of set1 and set2.

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print(set1)
print(set2)
```

```python
print(set1.union(set2))
print(set1.intersection(set2))
```

Performance Category Program:
    
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).

2. Find the performance category based on the input score using following criteria:
    
a. Above Average: Score greater than 7
    
b. Average: Score between 4 and 7(both inclusive)
    
c. Below Average: Score lesser than 4
    
3. Output: Print the Performance category
    
4. Additional Step: You can give a prompt of your choice to each category.
    
For eg: If score below average “Need to Improve your performance, consistent
practice will lead to better results”.

```python
score = int(input("Enter your score (0 to 10) : "))
if score>7 and score<=10:
    print("Above Average : Good performance.Keep up the good work")
elif score>=4 and score<=7:
    print("Average : satisfactory performance,It can improve with more practise")
elif score>=0 and score<4:
    print("Below Average : Needs more efforet & practise to improve")
else:
    print("Invalid Score")
```
